```python
#conditional statements
marks = int(input("enter the marks:"))
if marks>=90:
  print("grade A")
elif marks >= 80 and marks < 90:
  print("grade B")
elif marks >= 70 and marks < 80:
  print("grade C")
else:
  print("grade D")
```

```python
score = 85
attendance = 92

if score >= 60:
    # Outer condition passed (Student passed the exam)
    if attendance >= 90:
        print("Passed with Honors!")
    elif attendance >= 75:
        print("Passed.")
    else:
        print("Passed the exam, but failed due to low attendance.")
else:
    # Outer condition failed
    print("Failed the exam.")
```
